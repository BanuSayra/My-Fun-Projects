```python
# Asra Fun Game
# Number guessing game
from random import randint
guess_num = randint(1, 100)
print("Welcome to the Asra's Number Guessing Game!")
print("I'm thinking of a number between 1 and 100.")    
attempts = 0
while True:
    guess = int(input("Make a guess: "))
    attempts += 1
    if attempts > 10:
        print(f"You've used all your attempts. The number was {guess_num}.")
        break
    if guess < guess_num:
        print(" Low, guess Higher")
    elif guess > guess_num:
        print(" High, guess Lower")
    else:
        print(f"You got it! The answer was {guess_num}.")
        print(f"You guessed it in {attempts} attempts.")
        if attempts <= 3:
            print("Excellent work! Jackpot!")
        break
```
